```python
from docx import Document

# .docx 파일을 읽는 함수
def read_docx(file_name):
    file = os.path.join(file_path, file_name)
    doc = Document(file)
    content = []
    
    # 각 단락을 순회하며 텍스트를 가져옴
    for paragraph in doc.paragraphs:
        content.append(paragraph.text)
    
    return content

# 파일 경로를 지정하여 사용
file_name = "Seeds of Chaos.docx"
file_name_2 = "Seeds of Chaos_2.docx"
doc_content = read_docx(file_name) + read_docx(file_name_2)
```

```python
from docx import Document

# 새로운 .docx 파일을 생성하는 함수
def create_docx(file_path, content_list):
    # 새로운 Document 객체 생성
    doc = Document()
    
    # content_list에 있는 내용을 문서에 추가
    for content in content_list:
        doc.add_paragraph(content)
    
    # 지정된 경로에 .docx 파일 저장
    doc.save(file_path)

# 사용할 내용 리스트
content = [
    "첫 번째 문단입니다.",
    "두 번째 문단입니다.",
    "세 번째 문단입니다."
]

# 파일을 저장할 경로와 이름 설정
file_path = "output.docx"
create_docx(file_path, content)

print(f"{file_path} 파일이 생성되었습니다.")
```

```python
from docx import Document

def docx_token(file_name):
    file = os.path.join(file_path, file_name)
    doc = Document(file)
    token = []
    for paragraph in doc.paragraphs:
        text = paragraph.text
        text = text.replace("\"", "")
        
        word = [word.strip() for word in text.split(':')]
        if len(word) != 2:
            continue
        if word[0].strip() in token:
            continue
        else:
            token.append(word[0])
    token = sorted(token)
    for idx, i in enumerate(token, 1):  # 1부터 인덱스 시작
        if idx % 6 == 0:
            print(f"\"{i}\"", end=',\n')  # 6번째 항목마다 줄바꿈
        else:
            print(f"\"{i}\"", end=', ')  # 그 외는 공백으로 구분
        
# .docx 파일을 읽는 함수
def transfer_docx2txt(file_name):
    file = os.path.join(file_path, file_name)
    doc = Document(file)
    token = []
    content = []
    
    # 각 단락을 순회하며 텍스트를 가져옴
    for paragraph in doc.paragraphs:
        text = paragraph.text
        text = text.replace('내레이션', '나레이션')\
            .replace('화법', '나레이션')\
            .replace('설명', '나레이션')\
            .replace('남자 학생', '남학생')\
            .replace('남자학생', '남학생')\
            .replace('여자 학생', '여학생')\
            .replace('여자학생', '여학생')\
            .replace(' & ', '와(과) ')\
            .replace(' &', '와(과) ')\
            .replace('&', '와(과) ')\
            .replace('맨', '남자')\
            .replace('여성','여인')\
            .replace('여자아이', '소녀')\
            .replace('여자','여인')\
            .replace('고스트', '유령')\
            .replace('귀신', '유령')\
        
        
        text = text.replace('샌드 군인', '모래 군인')\
            .replace('? ???', '???')\
            .replace('Aciddrop', '애시드드롭')\
            .replace('B-Zone Prisoner', 'B구역')\
            .replace('B-Zone', 'B구역')\
            .replace('Barty', '벌티')\
            .replace('Blaze', '브라즈')\
            .replace('Ardent Lateran Citizen', '열렬한 라테론 시티')\
            .replace('Armorless Union', '알모레스 연합')\
            .replace('Bean the Infected', '감염된 콩')\
            .replace('교환수', '상담원')\
            .replace('교환자', '상담원')\
            .replace('오퍼레이터', '운영자')\
            .replace('Bitter Root', '쓴 뿌리')\
            .replace('Boatman', '뱃사공')\
            .replace('Prez', '프리즈')\
            .replace('Lateran', '라테라노')\
            .replace('Brisk', '활기찬')\
            .replace('활발한 라테라노 시민', '활기찬 라테라노 시민')\
            .replace('Candela', '켄델라')\
            .replace('Ceobe', '케보')\
            .replace('오믈렛라이스맨', '오믈렛 장사꾼')\
            .replace('오믈렛밥장수', '오믈렛 장사꾼')\
            .replace('인간 얼굴 돼지', '사람 얼굴을 한 돼지')\
            .replace('전 교장 선생님', '전 교장')\
            .replace('전 교장', '전 교장 선생님')\
            .replace('청과상', '청과물 상인')\
            .replace('청과물상인', '청과물 상인')\
            .replace('솔저', '군인')\
            .replace('코로로 씨', '코로로')\
            .replace('타카야 (남자)', '타카야')\
            .replace('히라', '히레')\
            .replace('히르', '히레')\
            .replace('히어', '히레')\
            # .replace('빨간 머리 남자', '빨간머리 남자')
                
        pass_word = [
            'BGM'
        ]

        if text in pass_word:
            continue
        
        word = [word.strip() for word in text.split(':')]
        if len(word) != 2:
            #print("word len is under 2")
            content[-1] += text
            continue
        else:
            content.append(text)
            
        if word[0].strip() in token:
            continue
        else:
            token.append(word[0])
    
    token = sorted(token)
    for idx, i in enumerate(token, 1):  # 1부터 인덱스 시작
        if idx % 6 == 0:
            print(i, end=',\n')  # 6번째 항목마다 줄바꿈
        else:
            print(i, end=', ')  # 그 외는 공백으로 구분
            
    save_path = os.path.join(file_path,'../novel_txt', file_name.replace('.docx', '.txt'))
    with open(save_path, 'w', encoding='utf-8') as file:
        for item in content:
            file.write(str(item) + '\n')  # 각 항목을 한 줄로 저장

file_name = "Arknights.docx"
# docx_token(file_name)
# transfer_docx2txt(file_name)
```
