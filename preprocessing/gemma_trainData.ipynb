```python
import json
import os

init_path = os.getcwd()

folder_path=os.path.join(init_path,"../dataset/9-nine- Episode 1/9-nine- Episode 1")
default_file_name = "9-nine- Episode 1_"
```

```python
for i in range(3):
    formal_file_name = default_file_name+f"{i+1}.json"
    informal_file_name = default_file_name+f"{i+1} copy.json"
    formal_file_path = os.path.join(folder_path, formal_file_name)
    informal_file_name = os.path.join(folder_path, informal_file_name)

    with open(formal_file_path, "r", encoding="utf-8") as json_data:
        formal_data = json.load(json_data)
    with open(informal_file_name, "r", encoding="utf-8") as json_data:
        informal_data = json.load(json_data)
    
```

```python
train_data=[]
for idx, dataset in enumerate(informal_data):
    for dataset_key in dataset.keys():
        for idx_2, data in enumerate(dataset[dataset_key]):
            key_1, value_1 = formal_data[idx][dataset_key][idx_2].popitem()
            key_2, value_2 = informal_data[idx][dataset_key][idx_2].popitem()
            if value_1 != value_2:
                word = {'text':f"<bos><start_of_turn>user\n{key_1}:{value_1}<end_of_turn>\n<start_of_turn>model\n{key_2}:{value_2}<end_of_turn><eos>"}
                train_data.append(word)
print(train_data[0:5])
print(len(train_data))
```
