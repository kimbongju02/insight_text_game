```python
import os
current_path = os.getcwd()
docx_file_path = os.path.join(current_path, '..\\dataset\\novel_docx')
txt_file_path = os.path.join(current_path, '..\\dataset\\visual-novels\\Parsed')

file_lists = os.listdir(txt_file_path)
```

```python
import os
import re
from docx import Document

# XML에서 허용되지 않는 문자들을 필터링하는 함수
def clean_text(text):
    # 제어 문자나 NULL 바이트 등을 제거
    return re.sub(r'[\x00-\x08\x0B\x0C\x0E-\x1F\x7F]', '', text)

for file_name in file_lists:
    file_path = os.path.join(txt_file_path, file_name)
    
    # 1. 텍스트 파일 읽기
    with open(file_path, "r", encoding="utf-8") as file:
        text = file.read()
    
    # 2. 읽은 텍스트를 클린 처리 (제어 문자 제거)
    cleaned_text = clean_text(text)

    # 3. .docx 파일 생성
    doc = Document()
    doc.add_paragraph(cleaned_text)

    # 4. .docx 파일 저장 경로 생성 및 저장
    save_docx_path = os.path.join(docx_file_path, file_name.replace('.txt','.docx'))
    doc.save(save_docx_path)

print("모든 txt 파일이 성공적으로 docx 파일로 변환되었습니다.")
```

```python
import re
import os
from docx import Document
def transfer_docx2txt(file_name):
    file = file_name
    doc = Document(file)
    tokens = []
    
    # 각 단락을 순회하며 텍스트를 가져옴
    for paragraph in doc.paragraphs:
        text = paragraph.text
    content = [t.strip(' \'\"') for t in text.split(',')]
        
    for token in content:
        if re.search(r'[a-zA-Z]', token) and not re.search(r'[\uAC00-\uD7A3]', token):
            tokens.append(token)

    return tokens

tokens = transfer_docx2txt("token_ko_v1.docx")
```
